```python
# ============================================================
#  CELL 3 · DATASET UPLOAD & EXPLORATORY ANALYSIS
# ============================================================

from google.colab import files as colab_files

TARGET_COLUMN = "sepsis_group"

# ── Column configuration for the NICU Sepsis dataset ─────────────────────
#  Columns to drop (IDs, outcomes, leakage, high-missingness)
DROP_COLS = [
    "episode_id", "unique_patient_id",          # IDs
    "race",                                      # sensitive attribute
    "period",                                    # time confound
    "cx_site",                                   # >90 % missing
    "onset_hour_of_day",                         # minimal clinical value
    "positive_days",                             # post-evaluation outcome
    "time_to_antibiotics", "stat_abx",           # treatment decisions (leakage)
    "intubated_free_days", "inotrope_free_days", # outcome free-days
    "overall_mortality_within_7_days",           # mortality outcomes (leakage)
    "overall_mortality_within_14_days",
    "overall_mortality_within_30_days",
    "length_of_stay_hours",                      # outcome (leakage)
]

# ── Selected clinical features (no data leakage) ─────────────────────────
SELECTED_FEATURES = [
    "sex",
    "gestational_age_at_birth_weeks",
    "birth_weight_kg",
    "temp_celsius",
    "onset_age_in_days",
    "blood_culture_positive",
    "intubated_at_time_of_sepsis_evaluation",
    "inotrope_at_time_of_sepsis_eval",
    "central_venous_line",
    "umbilical_arterial_line",
    "ecmo",
    "comorbidity_necrotizing_enterocolitis",
    "comorbidity_chronic_lung_disease",
    "comorbidity_cardiac",
    "comorbidity_surgical",
    "comorbidity_ivh_or_shunt",
]


def generate_synthetic():
    """Synthetic NICU dataset that mirrors real column structure."""
    np.random.seed(42)
    n = 900
    n1 = 194   # ~same ratio as real dataset

    def grp(n_s, hi):
        s = int(hi)
        return {
            "sex":                                   np.random.randint(0, 2, n_s),
            "gestational_age_at_birth_weeks":        np.clip(np.random.normal(34-2*s, 4, n_s), 22, 41),
            "birth_weight_kg":                       np.clip(np.random.normal(2.0-0.5*s, 0.7, n_s), 0.37, 5.95),
            "temp_celsius":                          np.clip(np.random.normal(37.3+0.8*s, 0.6, n_s), 33.7, 41.3),
            "onset_age_in_days":                     np.random.randint(0, 120+80*s, n_s),
            "blood_culture_positive":                np.random.choice([0,1], n_s, p=[0.6-0.3*s, 0.4+0.3*s]),
            "intubated_at_time_of_sepsis_evaluation":np.random.choice([0,1], n_s, p=[0.5-0.3*s, 0.5+0.3*s]),
            "inotrope_at_time_of_sepsis_eval":       np.random.choice([0,1], n_s, p=[0.7-0.35*s, 0.3+0.35*s]),
            "central_venous_line":                   np.random.choice([0,1], n_s, p=[0.45-0.2*s, 0.55+0.2*s]),
            "umbilical_arterial_line":               np.random.choice([0,1], n_s, p=[0.6-0.3*s, 0.4+0.3*s]),
            "ecmo":                                  np.random.choice([0,1], n_s, p=[0.97-0.1*s, 0.03+0.1*s]),
            "comorbidity_necrotizing_enterocolitis": np.random.choice([0,1], n_s, p=[0.85-0.3*s, 0.15+0.3*s]),
            "comorbidity_chronic_lung_disease":      np.random.choice([0,1], n_s, p=[0.55-0.2*s, 0.45+0.2*s]),
            "comorbidity_cardiac":                   np.random.choice([0,1], n_s, p=[0.8-0.2*s, 0.2+0.2*s]),
            "comorbidity_surgical":                  np.random.choice([0,1], n_s, p=[0.75-0.25*s, 0.25+0.25*s]),
            "comorbidity_ivh_or_shunt":              np.random.choice([0,1], n_s, p=[0.6-0.2*s, 0.4+0.2*s]),
            "sepsis_group":                          [1]*n_s if hi else [2]*n_s,
        }

    hi  = pd.DataFrame(grp(n1, True))
    lo  = pd.DataFrame(grp(n-n1, False))
    df  = pd.concat([hi, lo], ignore_index=True).sample(frac=1, random_state=42)
    print(f"✅ Synthetic dataset: {df.shape}")
    return df


def load_dataset():
    print("📂 Upload your CSV / Excel dataset (Cancel → synthetic fallback)...")
    try:
        uploaded = colab_files.upload()
        if not uploaded:
            raise ValueError("No file selected")
        fname = list(uploaded.keys())[0]
        ext   = os.path.splitext(fname)[1].lower()
        raw   = uploaded[fname]
        if ext == ".csv":
            df = pd.read_csv(io.BytesIO(raw))
        elif ext in (".xlsx", ".xlsm"):
            df = pd.read_excel(io.BytesIO(raw), engine="openpyxl")
        elif ext == ".xls":
            df = pd.read_excel(io.BytesIO(raw), engine="xlrd")
        else:
            raise ValueError(f"Unsupported extension: {ext}")
        print(f"✅ Loaded '{fname}'  |  {df.shape[0]} rows × {df.shape[1]} cols")
        return df
    except Exception as e:
        print(f"⚠️  Upload skipped ({e}).\n🔄 Using synthetic dataset...")
        return generate_synthetic()


RAW_DF = load_dataset()

# ── Quick EDA ─────────────────────────────────────────────────────────────
print("\n" + "="*55)
print("  DATASET SUMMARY")
print("="*55)
print(f"  Rows × Cols : {RAW_DF.shape[0]} × {RAW_DF.shape[1]}")
print(f"  Missing vals: {RAW_DF.isnull().sum().sum()} total")
if TARGET_COLUMN in RAW_DF.columns:
    print(f"\n  {TARGET_COLUMN} distribution:")
    print(RAW_DF[TARGET_COLUMN].value_counts().sort_index().to_string())
print("\n  Column list:")
for c in RAW_DF.columns:
    print(f"    {c:<45} {RAW_DF[c].dtype}")

RAW_DF.head(3)
```

```python
# ============================================================
#  CELL 4 · PREPROCESSING
#  - Binarise target (sepsis_group==1 → HIGH RISK)
#  - Drop leakage / ID columns
#  - Median / mode imputation
#  - StandardScaler (fit only on train, transform on test/predict)
#  - SMOTE class balancing (train only)
# ============================================================

# ── Global preprocessing objects ─────────────────────────────────────────
SCALER         = StandardScaler()
LABEL_ENCODERS = {}            # for any future categorical columns
FEATURE_COLUMNS = []           # ordered list — set after preprocessing
LE_TARGET       = LabelEncoder()


def binarise_target(series: pd.Series) -> pd.Series:
    """
    Map the multi-class sepsis_group to a binary risk label.
    Group 1 (confirmed blood-culture-positive sepsis) = HIGH RISK (1)
    All other groups (rule-out / uninfected / borderline) = LOW RISK (0)
    """
    return (series == 1).astype(int)


def select_features(df: pd.DataFrame) -> pd.DataFrame:
    """Return only the clinical features that exist in the dataframe."""
    available = [f for f in SELECTED_FEATURES if f in df.columns]
    missing   = [f for f in SELECTED_FEATURES if f not in df.columns]
    if missing:
        print(f"  ⚠️  Missing columns (will use zeros): {missing}")
    return df[available].copy()


def impute(df: pd.DataFrame) -> pd.DataFrame:
    """Median imputation for numeric, mode for categorical."""
    num_cols = df.select_dtypes(include=[np.number]).columns
    cat_cols = df.select_dtypes(include=["object", "category", "string"]).columns
    for c in num_cols:
        df[c] = df[c].fillna(df[c].median())
    for c in cat_cols:
        mode = df[c].mode()
        df[c] = df[c].fillna(mode[0] if len(mode) else "Unknown")
    return df


def encode_categorical(df: pd.DataFrame, fit: bool) -> pd.DataFrame:
    """Label-encode any remaining object/string columns."""
    cat_cols = df.select_dtypes(include=["object","category","string"]).columns.tolist()
    for c in cat_cols:
        if fit:
            le = LabelEncoder()
            df[c] = le.fit_transform(df[c].astype(str))
            LABEL_ENCODERS[c] = le
        else:
            le = LABEL_ENCODERS.get(c)
            if le:
                def safe(v, _le=le):
                    try:    return _le.transform([str(v)])[0]
                    except: return 0
                df[c] = df[c].astype(str).map(safe)
    return df


def preprocess(df: pd.DataFrame, fit: bool = True):
    """
    Full preprocessing pipeline.
    fit=True  → fit + transform  (training time)
    fit=False → transform only   (inference time)
    Returns (X_scaled, y_binary_or_None, feature_columns)
    """
    global SCALER, FEATURE_COLUMNS

    df = df.copy()

    # 1. Extract target
    y = None
    if TARGET_COLUMN in df.columns:
        y = binarise_target(df[TARGET_COLUMN])
        df = df.drop(columns=[TARGET_COLUMN])

    # 2. Drop irrelevant columns
    df = df.drop(columns=[c for c in DROP_COLS if c in df.columns], errors="ignore")

    # 3. Select feature subset
    X = select_features(df)

    # 4. Add any missing expected columns (as 0)
    for f in SELECTED_FEATURES:
        if f not in X.columns:
            X[f] = 0
    X = X[SELECTED_FEATURES]   # enforce column order

    # 5. Impute
    X = impute(X)

    # 6. Encode categorical
    X = encode_categorical(X, fit=fit)

    # 7. Store / align feature order
    if fit:
        FEATURE_COLUMNS = X.columns.tolist()
    else:
        X = X[FEATURE_COLUMNS]

    # 8. Scale
    X_arr = X.values.astype(float)
    X_sc  = SCALER.fit_transform(X_arr) if fit else SCALER.transform(X_arr)

    return X_sc, y, FEATURE_COLUMNS


# ── Run preprocessing ─────────────────────────────────────────────────────
print("⚙️  Running preprocessing pipeline...")
X_all, y_all, FEATURE_COLUMNS = preprocess(RAW_DF, fit=True)

print(f"  Features kept : {len(FEATURE_COLUMNS)}")
print(f"  Feature names : {FEATURE_COLUMNS}")
print(f"  X shape       : {X_all.shape}")
print(f"  Class counts  : {Counter(y_all)}")

# ── Train / test split (stratified 80/20) ────────────────────────────────
X_train, X_test, y_train, y_test = train_test_split(
    X_all, y_all, test_size=0.20, random_state=42, stratify=y_all
)
print(f"\n  Train : {X_train.shape[0]}  |  Test : {X_test.shape[0]}")
print(f"  Train class dist : {Counter(y_train)}")

# ── SMOTE (applied only to training set) ─────────────────────────────────
print("\n  Applying SMOTE to training set...")
smote = SMOTE(random_state=42, k_neighbors=5)
X_train_sm, y_train_sm = smote.fit_resample(X_train, y_train)
print(f"  After SMOTE : {Counter(y_train_sm)}")

print("\n✅ Preprocessing complete!")
```

```python
# ============================================================
#  CELL 5 · HYBRID STACKING MODEL
#
#  Architecture:
#    Base 1 : RandomForestClassifier    (200 trees, depth-capped)
#    Base 2 : MLPClassifier             (128→64→32, Adam, early stop)
#    Meta   : LogisticRegression        (trains on 5-fold OOF probs)
# ============================================================

print("🧠 Configuring base models...")

# ── Base Model 1: Random Forest ───────────────────────────────────────────
rf = RandomForestClassifier(
    n_estimators=250,
    max_depth=12,
    min_samples_split=4,
    min_samples_leaf=2,
    max_features="sqrt",
    class_weight="balanced",
    random_state=42,
    n_jobs=-1,
)

# ── Base Model 2: MLP Neural Network ─────────────────────────────────────
mlp = MLPClassifier(
    hidden_layer_sizes=(128, 64, 32),
    activation="relu",
    solver="adam",
    alpha=1e-4,                    # L2 regularisation
    learning_rate="adaptive",
    learning_rate_init=5e-4,
    max_iter=600,
    early_stopping=True,
    validation_fraction=0.10,
    n_iter_no_change=20,
    random_state=42,
)

# ── Meta Learner: Logistic Regression ────────────────────────────────────
meta = LogisticRegression(
    C=0.5,
    solver="lbfgs",
    max_iter=1000,
    class_weight="balanced",
    random_state=42,
)

# ── Stacking Classifier ───────────────────────────────────────────────────
STACKING_MODEL = StackingClassifier(
    estimators=[
        ("random_forest",  rf),
        ("neural_network", mlp),
    ],
    final_estimator=meta,
    cv=StratifiedKFold(n_splits=5, shuffle=True, random_state=42),
    stack_method="predict_proba",
    passthrough=False,             # meta-learner sees only base predictions
    n_jobs=1,                      # -1 can cause issues in Colab
)

print("  Base-1 : RandomForestClassifier  (n=250, depth≤12, balanced)")
print("  Base-2 : MLPClassifier            (128→64→32, Adam, early stop)")
print("  Meta   : LogisticRegression       (5-fold stratified CV)")
print("\n⏳ Training hybrid stacking model...")
print("   (Expected time: 2–5 minutes on Colab CPU)\n")

import time
t0 = time.time()
STACKING_MODEL.fit(X_train_sm, y_train_sm)
elapsed = time.time() - t0
print(f"✅ Training complete in {elapsed:.1f}s")

# ── Evaluation ───────────────────────────────────────────────────────────
y_pred       = STACKING_MODEL.predict(X_test)
y_pred_proba = STACKING_MODEL.predict_proba(X_test)[:, 1]

MODEL_METRICS = {
    "Accuracy":  round(accuracy_score(y_test, y_pred) * 100, 2),
    "Precision": round(precision_score(y_test, y_pred, zero_division=0) * 100, 2),
    "Recall":    round(recall_score(y_test, y_pred, zero_division=0) * 100, 2),
    "F1-Score":  round(f1_score(y_test, y_pred, zero_division=0) * 100, 2),
    "ROC-AUC":   round(roc_auc_score(y_test, y_pred_proba) * 100, 2),
}

print("\n" + "═"*50)
print("  TEST-SET PERFORMANCE METRICS")
print("═"*50)
for k, v in MODEL_METRICS.items():
    bar = "█" * int(v/5) + "░" * (20 - int(v/5))
    print(f"  {k:<12} {bar} {v:.2f}%")
print("═"*50)

print("\nDetailed Classification Report:")
print(classification_report(
    y_test, y_pred,
    target_names=["Low Risk (0)", "High Risk (1)"],
    digits=4
))
```

```python
# ============================================================
#  CELL 6 · EVALUATION PLOTS
# ============================================================

PLOT_DIR = "/tmp/plots"
os.makedirs(PLOT_DIR, exist_ok=True)

PALETTE = {"high":"#C0392B", "low":"#1A7B4B",
           "blue":"#1565C0", "ltblue":"#AED6F1",
           "header":"#1A3A5C", "bg":"#F0F8FF"}


def save_plot(fig, name):
    path = f"{PLOT_DIR}/{name}.png"
    fig.savefig(path, dpi=160, bbox_inches="tight",
                facecolor="white", edgecolor="none")
    plt.show()
    plt.close(fig)
    return path


# ── 1. Confusion Matrix ───────────────────────────────────────────────────
def plot_cm():
    cm  = confusion_matrix(y_test, y_pred)
    fig, ax = plt.subplots(figsize=(5.2, 4.2))
    disp = ConfusionMatrixDisplay(cm, display_labels=["Low Risk","High Risk"])
    disp.plot(ax=ax, colorbar=False, cmap="Blues")
    ax.set_title("Confusion Matrix — Stacking Classifier",
                 fontsize=12, fontweight="bold", pad=10, color=PALETTE["header"])
    ax.set_xlabel("Predicted Label", fontsize=10)
    ax.set_ylabel("True Label",      fontsize=10)
    for text in ax.texts:
        text.set_fontsize(16)
    fig.tight_layout()
    return save_plot(fig, "confusion_matrix")

# ── 2. ROC Curve ─────────────────────────────────────────────────────────
def plot_roc():
    fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
    auc  = roc_auc_score(y_test, y_pred_proba)
    fig, ax = plt.subplots(figsize=(5.5, 4.5))
    ax.plot(fpr, tpr, color=PALETTE["blue"], lw=2.5,
            label=f"Stacking Model  (AUC = {auc:.4f})")
    ax.fill_between(fpr, tpr, alpha=0.12, color=PALETTE["blue"])
    ax.plot([0,1],[0,1], "k--", lw=1.3, alpha=0.55, label="Random Classifier")
    ax.set_xlabel("False Positive Rate", fontsize=11)
    ax.set_ylabel("True Positive Rate",  fontsize=11)
    ax.set_title("ROC-AUC Curve — Neonatal Sepsis Detection",
                 fontsize=12, fontweight="bold", pad=10, color=PALETTE["header"])
    ax.legend(loc="lower right", fontsize=10, framealpha=0.9)
    ax.grid(True, alpha=0.25, linestyle="--")
    ax.set_facecolor(PALETTE["bg"])
    fig.tight_layout()
    return save_plot(fig, "roc_curve")

# ── 3. Feature Importance (Random Forest) ────────────────────────────────
def plot_fi():
    rf_est  = STACKING_MODEL.named_estimators_["random_forest"]
    imp     = rf_est.feature_importances_
    idx     = np.argsort(imp)[::-1][:12]
    labels  = [FEATURE_COLUMNS[i].replace("comorbidity_","comrb_")
                                  .replace("_at_time_of_sepsis_evaluation","")
                                  .replace("_at_time_of_sepsis_eval","")
                                  .replace("_at_birth_weeks"," (wks)")
               for i in idx]
    vals    = imp[idx]
    fig, ax = plt.subplots(figsize=(7.5, 4.5))
    palette = sns.color_palette("Blues_r", len(labels))
    bars    = ax.barh(labels[::-1], vals[::-1], color=palette,
                      edgecolor="white", linewidth=0.5)
    for bar, v in zip(bars, vals[::-1]):
        ax.text(v + 0.002, bar.get_y() + bar.get_height()/2,
                f"{v:.4f}", va="center", fontsize=8.5, color="#333")
    ax.set_xlabel("Gini Importance", fontsize=11)
    ax.set_title("Feature Importance — Random Forest Base Model",
                 fontsize=12, fontweight="bold", pad=10, color=PALETTE["header"])
    ax.spines[["top","right"]].set_visible(False)
    ax.set_facecolor(PALETTE["bg"])
    fig.tight_layout()
    return save_plot(fig, "feature_importance")

# ── 4. Probability Distribution ──────────────────────────────────────────
def plot_prob_dist():
    fig, ax = plt.subplots(figsize=(5.5, 4.0))
    proba_hi = y_pred_proba[y_test == 1]
    proba_lo = y_pred_proba[y_test == 0]
    ax.hist(proba_lo, bins=30, color=PALETTE["low"],  alpha=0.65,
            label="Low Risk (True)", edgecolor="white")
    ax.hist(proba_hi, bins=30, color=PALETTE["high"], alpha=0.65,
            label="High Risk (True)", edgecolor="white")
    ax.axvline(0.5, color="#333", linestyle="--", lw=1.5, label="Threshold 0.5")
    ax.set_xlabel("Predicted Sepsis Probability", fontsize=11)
    ax.set_ylabel("Frequency",                   fontsize=11)
    ax.set_title("Predicted Probability Distribution",
                 fontsize=12, fontweight="bold", pad=8, color=PALETTE["header"])
    ax.legend(fontsize=9, framealpha=0.9)
    ax.set_facecolor(PALETTE["bg"])
    fig.tight_layout()
    return save_plot(fig, "prob_dist")

print("📊 Generating evaluation plots...")
CM_PATH   = plot_cm()
ROC_PATH  = plot_roc()
FI_PATH   = plot_fi()
PROB_PATH = plot_prob_dist()
print("\n✅ All plots saved.")
```

```python
# ============================================================
#  CELL 9 · INFERENCE ENGINE
#  Maps UI inputs → model features → prediction
#  Mirrors the exact preprocessing used during training.
# ============================================================

def build_feature_row(patient: dict) -> np.ndarray:
    """
    Convert a patient dict (from UI) into a scaled feature array
    using the same scaler and feature order as training.
    """
    # Map patient dict keys → dataset column names
    row = {
        "sex":                                    patient.get("sex_code", 1),
        "gestational_age_at_birth_weeks":         patient.get("gestational_age", 32.0),
        "birth_weight_kg":                        patient.get("birth_weight", 1.5),
        "temp_celsius":                           patient.get("temperature", 37.3),
        "onset_age_in_days":                      patient.get("onset_age_days", 44),
        "blood_culture_positive":                 patient.get("blood_culture_positive", 0),
        "intubated_at_time_of_sepsis_evaluation": patient.get("intubated", 0),
        "inotrope_at_time_of_sepsis_eval":        patient.get("inotrope", 0),
        "central_venous_line":                    patient.get("central_venous_line", 0),
        "umbilical_arterial_line":                patient.get("umbilical_arterial_line", 0),
        "ecmo":                                   patient.get("ecmo", 0),
        "comorbidity_necrotizing_enterocolitis":  patient.get("comrb_nec", 0),
        "comorbidity_chronic_lung_disease":       patient.get("comrb_lung", 0),
        "comorbidity_cardiac":                    patient.get("comrb_cardiac", 0),
        "comorbidity_surgical":                   patient.get("comrb_surgical", 0),
        "comorbidity_ivh_or_shunt":               patient.get("comrb_ivh", 0),
    }

    # Build DataFrame aligned to training columns
    df_row = pd.DataFrame([{f: row.get(f, 0) for f in FEATURE_COLUMNS}])
    df_row = df_row[FEATURE_COLUMNS].astype(float)

    # Apply the same scaler (already fitted on training data)
    return SCALER.transform(df_row.values)


def predict_sepsis(patient: dict):
    """
    Run stacking model on a patient dict.
    Returns (label, probability_pct, risk_level).
    """
    X_sc       = build_feature_row(patient)
    pred_cls   = STACKING_MODEL.predict(X_sc)[0]
    pred_proba = STACKING_MODEL.predict_proba(X_sc)[0]
    prob_high  = round(float(pred_proba[1]) * 100, 2)
    risk_level = "High Risk" if pred_cls == 1 else "Low Risk"
    label      = "Sepsis Detected" if pred_cls == 1 else "No Sepsis"
    return label, prob_high, risk_level


# ── Sanity check ─────────────────────────────────────────────────────────
_hi = predict_sepsis({
    "sex_code":1, "gestational_age":26.0, "birth_weight":0.8,
    "temperature":39.2, "onset_age_days":12,
    "blood_culture_positive":1, "intubated":1, "inotrope":1,
    "central_venous_line":1, "umbilical_arterial_line":1, "ecmo":0,
    "comrb_nec":1, "comrb_lung":1, "comrb_cardiac":0,
    "comrb_surgical":0, "comrb_ivh":1,
})
_lo = predict_sepsis({
    "sex_code":0, "gestational_age":38.0, "birth_weight":3.2,
    "temperature":37.1, "onset_age_days":5,
    "blood_culture_positive":0, "intubated":0, "inotrope":0,
    "central_venous_line":0, "umbilical_arterial_line":0, "ecmo":0,
    "comrb_nec":0, "comrb_lung":0, "comrb_cardiac":0,
    "comrb_surgical":0, "comrb_ivh":0,
})
print("✅ Inference engine ready!")
print(f"  High-risk patient → {_hi[2]} | prob {_hi[1]:.1f}%")
print(f"  Low-risk  patient → {_lo[2]} | prob {_lo[1]:.1f}%")
```

```python
# ============================================================
#  CELL 11 · GRADIO WEB APPLICATION
#  Compatible with gradio 3.50.2 — no HfFolder dependency.
# ============================================================

CSS = """
/* ─── Root variables ──────────────────────────────────── */
body, .gradio-container {
  background: linear-gradient(135deg, #EBF5FB 0%, #D6EAF8 100%) !important;
  font-family: 'Segoe UI', 'Helvetica Neue', sans-serif !important;
}
/* ─── Header ──────────────────────────────────────────── */
.app-hdr {
  background: linear-gradient(100deg, #1A3A5C 0%, #2C5282 55%, #1A6FA0 100%);
  padding: 22px 28px 18px; border-radius: 12px;
  box-shadow: 0 4px 18px rgba(26,58,92,.30); margin-bottom: 6px;
}
.app-hdr h1 {
  color: #fff !important; font-size: 1.45rem; font-weight: 700;
  margin: 0 0 5px; text-shadow: 0 1px 5px rgba(0,0,0,.35);
}
.app-hdr p { color: #A9CCE3 !important; font-size: .86rem; margin: 0; }
/* ─── Section labels ──────────────────────────────────── */
.sect-lbl {
  font-size: .75rem; font-weight: 700; color: #1A3A5C;
  text-transform: uppercase; letter-spacing: .08em;
  border-bottom: 2px solid #2196F3; padding-bottom: 4px;
  margin: 12px 0 10px;
}
/* ─── Labels ──────────────────────────────────────────── */
label, .gr-form label { color: #1A3A5C !important; font-weight: 600 !important;
  font-size: .82rem !important; }
/* ─── Inputs ──────────────────────────────────────────── */
input[type=number], input[type=text], select, textarea {
  border: 1.5px solid #AED6F1 !important; border-radius: 7px !important;
  font-size: .9rem !important; background: #fff !important;
  transition: border-color .18s, box-shadow .18s;
}
input:focus, select:focus, textarea:focus {
  border-color: #2196F3 !important;
  box-shadow: 0 0 0 3px rgba(33,150,243,.13) !important;
}
/* ─── Predict button ──────────────────────────────────── */
#pred-btn {
  background: linear-gradient(90deg, #1A3A5C, #2C5282) !important;
  color: #fff !important; border: none !important; border-radius: 8px !important;
  font-size: .97rem !important; font-weight: 700 !important;
  padding: 12px 0 !important; width: 100% !important;
  box-shadow: 0 3px 12px rgba(26,58,92,.30); letter-spacing: .04em;
  cursor: pointer; transition: all .22s;
}
#pred-btn:hover {
  background: linear-gradient(90deg, #2C5282, #1A6FA0) !important;
  box-shadow: 0 6px 20px rgba(26,58,92,.40); transform: translateY(-1px);
}
/* ─── Output textbox ──────────────────────────────────── */
#res-box textarea {
  font-family: 'Courier New', monospace !important;
  font-size: .88rem !important; line-height: 1.7 !important;
  background: #F0F8FF !important; border: 1.5px solid #AED6F1 !important;
  border-radius: 10px !important; color: #1C2833 !important;
}
/* ─── Metric chips ────────────────────────────────────── */
.chip {
  display: inline-block; background: #EBF5FB;
  border: 1.5px solid #AED6F1; border-radius: 8px;
  padding: 5px 13px; margin: 3px; font-weight: 600;
  color: #1A3A5C; font-size: .84rem;
}
/* ─── Table ───────────────────────────────────────────── */
table { border-collapse: collapse; width: 100%; font-size: .82rem; }
th { background: #1A3A5C; color: #fff; padding: 7px 10px; text-align: left; }
td { padding: 6px 10px; border-bottom: 1px solid #D6EAF8; color: #1C2833; }
tr:nth-child(even) td { background: #F0F8FF; }
"""


# ── Prediction handler ─────────────────────────────────────────────────────
def handle_predict(
    patient_name, blood_group, sex_ui,
    gestational_age, birth_weight, temperature, onset_age,
    blood_cx, intubated, inotrope, cvl, ual, ecmo,
    nec, cld, cardiac, surgical, ivh
):
    try:
        sex_code  = 1 if sex_ui == "Male" else 0
        patient   = {
            "patient_name":         patient_name.strip() or "Anonymous",
            "blood_group":          blood_group,
            "sex":                  sex_ui,
            "sex_code":             sex_code,
            "gestational_age":      float(gestational_age),
            "birth_weight":         float(birth_weight),
            "temperature":          float(temperature),
            "onset_age_days":       int(onset_age),
            "blood_culture_positive":int(blood_cx),
            "intubated":            int(intubated),
            "inotrope":             int(inotrope),
            "central_venous_line":  int(cvl),
            "umbilical_arterial_line":int(ual),
            "ecmo":                 int(ecmo),
            "comrb_nec":            int(nec),
            "comrb_lung":           int(cld),
            "comrb_cardiac":        int(cardiac),
            "comrb_surgical":       int(surgical),
            "comrb_ivh":            int(ivh),
        }

        label, prob, risk = predict_sepsis(patient)
        rec_id            = save_record(patient, label, prob, risk)
        pdf_path          = generate_report(patient, label, prob, risk, rec_id)

        ico   = "🔴" if risk == "High Risk" else "🟢"
        bar   = "█" * int(prob/5) + "░" * (20 - int(prob/5))
        yn    = lambda v: "Yes" if int(v)==1 else "No"

        comorbs = []
        if int(nec):     comorbs.append("NEC")
        if int(cld):     comorbs.append("Chronic Lung Disease")
        if int(cardiac): comorbs.append("Cardiac")
        if int(surgical):comorbs.append("Surgical")
        if int(ivh):     comorbs.append("IVH/Shunt")
        cmrb_str = ", ".join(comorbs) if comorbs else "None"

        result = (
            f"{'═'*52}\n"
            f"   NEONATAL SEPSIS RISK ASSESSMENT\n"
            f"{'═'*52}\n\n"
            f"  {ico}  {risk.upper()}\n"
            f"  Outcome  : {label}\n\n"
            f"  Sepsis Probability:\n"
            f"  {bar}\n"
            f"  {' '*2}{prob:.1f}% (High-Risk Confidence)\n\n"
            f"{'─'*52}\n"
            f"  PATIENT DETAILS\n"
            f"{'─'*52}\n"
            f"  Name      : {patient['patient_name']}\n"
            f"  Blood Grp : {blood_group}   Sex: {sex_ui}\n"
            f"  Gest. Age : {gestational_age} wks   "
            f"Weight: {birth_weight} kg\n"
            f"  Temp      : {temperature} °C   "
            f"Onset: Day {int(onset_age)}\n\n"
            f"{'─'*52}\n"
            f"  CLINICAL STATUS\n"
            f"{'─'*52}\n"
            f"  Blood Culture : {'Positive' if int(blood_cx) else 'Negative'}\n"
            f"  Intubated     : {yn(intubated)}   "
            f"Inotrope : {yn(inotrope)}\n"
            f"  CVL           : {yn(cvl)}   "
            f"UAL      : {yn(ual)}\n"
            f"  ECMO          : {yn(ecmo)}\n"
            f"  Comorbidities : {cmrb_str}\n\n"
            f"{'─'*52}\n"
            f"  Record ID  : #{rec_id}\n"
            f"  Timestamp  : "
            f"{datetime.datetime.now().strftime('%Y-%m-%d %H:%M:%S')}\n\n"
            f"{'═'*52}\n"
            f"  ⚕  All findings must be reviewed by a physician.\n"
            f"{'═'*52}"
        )
        return result, pdf_path

    except Exception as exc:
        import traceback
        err = traceback.format_exc()
        return f"❌ Prediction Error:\n\n{err}", None


def refresh_hist():
    df = fetch_records(20)
    if df.empty:
        return "<p style='padding:12px;color:#5D6D7E'>No predictions yet.</p>"
    show = ["id","timestamp","patient_name","blood_group","sex",
            "prediction_label","sepsis_probability_pct","risk_level"]
    df2  = df[[c for c in show if c in df.columns]]
    return "<div style='overflow-x:auto'>" + df2.to_html(index=False) + "</div>"


def metrics_panel():
    chips = "".join(f"<span class='chip'>{k}: {v:.2f}%</span>"
                    for k, v in MODEL_METRICS.items())
    return (
        "<div style='background:#fff;border:1px solid #AED6F1;"
        "border-radius:10px;padding:18px'>"
        "<div class='sect-lbl'>Model Performance — Held-Out Test Set</div>"
        f"{chips}"
        "<p style='font-size:.8rem;color:#5D6D7E;margin-top:10px'>"
        "Architecture: RandomForest + MLPClassifier → LogisticRegression (5-fold CV) "
        "&nbsp;·&nbsp; SMOTE balancing &nbsp;·&nbsp; "
        "Target: sepsis_group==1 (confirmed) vs rest</p></div>"
    )


# ── Gradio Blocks UI ───────────────────────────────────────────────────────
with gr.Blocks(css=CSS, title="NICU Sepsis Predictor") as demo:

    gr.HTML("""
    <div class="app-hdr">
      <h1>🏥 Late-Onset Neonatal Sepsis Prediction System</h1>
      <p>Hybrid Stacking ML Model &nbsp;·&nbsp;
         RandomForest + MLP Neural Network → Logistic Regression
         &nbsp;·&nbsp; SMOTE · SQLite · PDF Reports</p>
    </div>
    """)

    with gr.Tabs():

        # ══ Tab 1: Prediction ══════════════════════════════════════════
        with gr.TabItem("🔬 Predict Sepsis Risk"):
            with gr.Row():

                # ── Left panel: inputs ────────────────────────────────
                with gr.Column():
                    gr.HTML("<div class='sect-lbl'>Patient Details</div>")
                    patient_name = gr.Textbox(
                        label="Mother / Guardian Name",
                        placeholder="e.g. Mrs. Anjali Nair",
                        value="")
                    blood_group = gr.Dropdown(
                        label="Baby's Blood Group",
                        choices=["A+","A-","B+","B-","AB+","AB-","O+","O-"],
                        value="O+")
                    sex_ui = gr.Dropdown(
                        label="Sex of Baby",
                        choices=["Male","Female"],
                        value="Male")

                    gr.HTML("<div class='sect-lbl'>Neonatal Measurements</div>")
                    gestational_age = gr.Number(
                        label="Gestational Age at Birth (weeks)",
                        value=32, precision=1)
                    birth_weight = gr.Number(
                        label="Birth Weight (kg)",
                        value=1.5, precision=2)
                    temperature = gr.Number(
                        label="Body Temperature (°C)",
                        value=37.3, precision=1)
                    onset_age = gr.Number(
                        label="Age at Sepsis Evaluation (days)",
                        value=44, precision=0)

                    gr.HTML("<div class='sect-lbl'>Clinical Indicators</div>")
                    blood_cx = gr.Radio(
                        label="Blood Culture Positive",
                        choices=[("Yes",1),("No",0)], value=0)
                    intubated = gr.Radio(
                        label="Intubated at Evaluation",
                        choices=[("Yes",1),("No",0)], value=0)
                    inotrope = gr.Radio(
                        label="Inotrope at Evaluation",
                        choices=[("Yes",1),("No",0)], value=0)
                    cvl = gr.Radio(
                        label="Central Venous Line",
                        choices=[("Yes",1),("No",0)], value=0)
                    ual = gr.Radio(
                        label="Umbilical Arterial Line",
                        choices=[("Yes",1),("No",0)], value=0)
                    ecmo = gr.Radio(
                        label="ECMO",
                        choices=[("Yes",1),("No",0)], value=0)

                    gr.HTML("<div class='sect-lbl'>Comorbidities</div>")
                    nec     = gr.Checkbox(label="Necrotizing Enterocolitis (NEC)", value=False)
                    cld     = gr.Checkbox(label="Chronic Lung Disease",            value=False)
                    cardiac = gr.Checkbox(label="Cardiac Comorbidity",             value=False)
                    surgical= gr.Checkbox(label="Surgical Comorbidity",            value=False)
                    ivh     = gr.Checkbox(label="IVH or Shunt",                   value=False)

                    pred_btn = gr.Button(
                        "⚕  Analyse & Predict Sepsis Risk",
                        elem_id="pred-btn")

                # ── Right panel: output ────────────────────────────────
                with gr.Column():
                    gr.HTML("<div class='sect-lbl'>Risk Assessment Result</div>")
                    result_out = gr.Textbox(
                        label="Prediction Output",
                        lines=26, interactive=False,
                        elem_id="res-box",
                        placeholder="Fill in patient details and click\n"
                                    "'Analyse & Predict Sepsis Risk' to see results.")
                    pdf_out = gr.File(
                        label="📄 Download PDF Report",
                        interactive=False)

            pred_btn.click(
                fn=handle_predict,
                inputs=[patient_name, blood_group, sex_ui,
                        gestational_age, birth_weight, temperature, onset_age,
                        blood_cx, intubated, inotrope, cvl, ual, ecmo,
                        nec, cld, cardiac, surgical, ivh],
                outputs=[result_out, pdf_out]
            )

        # ══ Tab 2: Model Performance ═══════════════════════════════════
        with gr.TabItem("📊 Model Performance"):
            gr.HTML(metrics_panel())
            with gr.Row():
                gr.Image(CM_PATH,   label="Confusion Matrix",  type="filepath")
                gr.Image(ROC_PATH,  label="ROC-AUC Curve",     type="filepath")
            with gr.Row():
                gr.Image(FI_PATH,   label="Feature Importance (Random Forest)", type="filepath")
                gr.Image(PROB_PATH, label="Probability Distribution",           type="filepath")

        # ══ Tab 3: History ═════════════════════════════════════════════
        with gr.TabItem("🗃️ Prediction History"):
            ref_btn  = gr.Button("🔄 Refresh Records")
            hist_out = gr.HTML(value=refresh_hist())
            ref_btn.click(fn=refresh_hist, inputs=[], outputs=hist_out)

        # ══ Tab 4: About ═══════════════════════════════════════════════
        with gr.TabItem("ℹ️ About"):
            gr.HTML(f"""
            <div style='background:#fff;border:1px solid #AED6F1;
                        border-radius:10px;padding:22px;line-height:1.75'>
              <div class='sect-lbl'>About This System</div>
              <p><b>Project:</b> A Robust Hybrid Stacking Architecture for
                 Late-Onset Neonatal Sepsis Prediction Using NICU Clinical Data</p>
              <b>Model Architecture:</b>
              <ul>
                <li><b>Base Model 1</b> — RandomForestClassifier
                    (250 trees · max_depth=12 · class_weight=balanced)</li>
                <li><b>Base Model 2</b> — MLPClassifier
                    (128→64→32 · ReLU · Adam · early stopping)</li>
                <li><b>Meta-Learner</b> — LogisticRegression
                    (trained on 5-fold stratified cross-val OOF probabilities)</li>
              </ul>
              <b>Preprocessing Pipeline:</b>
              ID removal → leakage exclusion → median/mode imputation →
              feature selection (16 clinical variables) → StandardScaler →
              80/20 stratified split → SMOTE on training fold only<br><br>
              <b>Target Definition:</b> sepsis_group == 1
              (confirmed blood-culture-positive sepsis) → HIGH RISK;
              all other groups → LOW RISK<br><br>
              <b>Test-Set Performance:</b><br><br>
              {"".join(f"<span class='chip'>{k}: {v:.2f}%</span>"
                       for k, v in MODEL_METRICS.items())}
              <br><br>
              <div style='background:#FEF9E7;padding:11px;border-radius:7px;
                          border-left:3px solid #F39C12;font-size:.83rem'>
                ⚠️ <b>Disclaimer:</b> This tool is intended for research and
                educational purposes only. Predictions must be reviewed by a
                qualified neonatologist before any clinical decision is made.
              </div>
            </div>
            """)


# ── Launch ─────────────────────────────────────────────────────────────────
print("\n" + "═"*60)
print("  🚀 LAUNCHING GRADIO APPLICATION")
print("  gradio", gr.__version__, "— no HfFolder dependency")
print("═"*60)
print("  A public shareable link will appear below.")
print("  The link is active for 72 hours.")
print("═"*60 + "\n")

demo.launch(
    share=True,        # creates public tunnel
    debug=False,
    show_error=True,
    height=950,
    quiet=False,
)
```
